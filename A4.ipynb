## 1. Motivation and Problem Statement

During the COVID-19 pandemic, people’s views on vaccines may have been shaped partly by what they read and discussed online. This project looks at one pattern in a public vaccine forum: 

**When someone comments on a post, does the comment usually take the same side on vaccines (for or against) as the post?**

This question is scientifically interesting because most studies of vaccine discussion on Reddit measure the overall mix of opinion. Pairing each comment with its post goes further: it shows how opinions are arranged, and whether people mostly respond within their own side or across sides. The results may also have practical value. If skeptical posts mostly draw skeptical comments, the forum may reinforce doubts more than it exposes hesitant people to other views, and knowing this helps health communicators and moderators judge where joining these conversations could reach people who are unsure. 

From a human-centered perspective, each row is a real person’s health opinion compressed into a single label, usually by a model rather than a person, so a worried question or a sarcastic remark can easily be mislabeled or overlooked. This matters because findings like “skeptical posts draw skeptical replies” could shape how moderators and health communicators treat hesitant users, and conclusions built on mislabeled opinions could lead them to misjudge or dismiss those people. I will present the results as patterns in labels rather than facts about people’s beliefs, and state clearly where the labels are uncertain and whose voices are missing.

I hope to learn whether comments mostly take the same side as the post they answer, compared with what the overall mix of sides would predict by chance. I also want to know whether this pattern differs for quick comments, posted within 6 hours of the post, compared with later ones. I chose this cutoff in advance. To answer these questions, I will pair each comment with its post, keep pairs where both messages took a side, and summarize them with counts, percentages, and simple charts.

## 2. Data Selected for Analysis

I will use Brambilla & Kharmale (2022), COVID-19 Vaccine Discussions on Reddit with Sentiment, Stance, Topics, and Timing, published on Harvard Dataverse at https://doi.org/10.7910/DVN/XJTBQM under a Creative Commons Attribution 4.0 (CC BY 4.0) license, which allows reuse with credit. The dataset contains 1,727 posts and 12,578 comments from r/CovidVaccine, posted between April 2020 and May 2021, and was built for the authors’ study of how Reddit comments relate to the discussions they belong to (Brambilla et al., 2022). 

Each message carries sentiment scores, topics, and a stance label (in favor, against, or none/neutral), and each comment records the post it belongs to and a time category for how long after the post it appeared. The authors hand-labeled stance for only 249 messages and built a model to predict it for the rest; the model labeled most comments as having no stance, about 2,000 as in favor, and about 800 as against. The dataset fits my question because it combines stance labels, the link between comments and posts, and timing. It also leaves room for new work: the original paper compared comments with their starting post for sentiment and topic, but not for stance.

This dataset raises ethical considerations. The posts are public, but the people who wrote them did not agree to be studied. Their health opinions are personal, and quoted text can often be traced back to its author through search. For these reasons, I will report only aggregate results, will not quote or name individual users, and will not republish raw comment text, even though the license would allow it. Because nearly all stance labels come from a model trained on a small hand-labeled set, they may misread sarcasm or mixed views. I also noted that the data represents only English-speaking users of one subreddit, not the public in general.

## 3. Unknowns and dependencies

Several factors outside of my control could affect this project. While the dataset is licensed CC BY 4.0, which allows reuse with credit, the comments themselves were written by Reddit users under Reddit’s own terms. Thus, I will publish only my code and aggregate results, not the raw comment text.

Next, most comments are labeled as having no stance, partly because the authors’ rules label a comment as for or against only when it states its view explicitly, so even a reply that simply agrees with its post counts as having no stance. I can therefore study only the smaller set of comments that took a clear side, and because nearly all labels were predicted by a model whose rules I can only partly see, I will treat them as approximate. 

Third, labeled pairs become rare in the days after a post, so I will compare only two time groups. Lastly, Reddit’s ranking affects which comments people see and answer, and I cannot observe or control it. 

I will stick to counts, percentages, and simple charts. So far, I have counted the usable pairs to check feasibility, but I have not yet tested the timing question.
